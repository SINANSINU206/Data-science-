```python
def replace_outlier(dataframe, feature):
    q1 = dataframe[feature].quantile(0.25)
    q3 = dataframe[feature].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    higher = q3 + 1.5 * iqr
    dataframe[feature] = np.where(
    dataframe[feature] < lower,
    lower,
    dataframe[feature]
    )
    dataframe[feature] = np.where(
        dataframe[feature] > higher,
        higher,
        dataframe[feature]
    )
```
